# Matrices, matrix expressions and N-dim arrays

Everything that SymPy's LaTeX printer can render is editable: concrete matrices (dense and sparse), `MatrixSymbol` algebra, block matrices, determinants/traces and `ImmutableDenseNDimArray`s of any rank.  Entries are addressable individually; editing an entry rebuilds the container.

```python
from sympy import *
from sympy.tensor.array import derive_by_array, tensorproduct
from sympy_editor import edit, register_op

x, y, z, t, theta = symbols("x y z t theta")
A, B = MatrixSymbol("A", 2, 2), MatrixSymbol("B", 2, 2)
```

## Concrete matrices

Click an entry, type a replacement, press Enter.  Select an entry and use **Apply** to transform just that entry.

```python
M = edit(Matrix([[x, y], [z, t]]))
M
```

```python
M.expr.det()
```

Matrix-valued ops are easy to add; they show up in the dropdown of widgets created afterwards (and report an error if applied to a non-matrix selection).

```python
register_op("transpose", lambda e: e.T, label="Transpose")
register_op("inverse", lambda e: e.inv() if hasattr(e, "inv") else e.I, label="Inverse")
register_op("det", lambda e: e.det(), label="Determinant")

R = Matrix([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]])
w = edit(R * Matrix([x, y]))   # select the whole product (↑ to the root) and Apply 'Evaluate (doit)'
w
```

## Matrix expressions

`MatrixSymbol` trees (`MatAdd`, `MatMul`, `Transpose`, `Inverse`, ...) are edited structurally like any other expression.

```python
edit(A * B + 2 * A.T - A.I)
```

```python
edit(BlockMatrix([[A, B], [B, A]]))
```

```python
edit(Determinant(Matrix([[x, y], [z, t]])) + Trace(A))   # select the determinant, Apply 'Evaluate (doit)'
```

## N-dimensional arrays

`Array` objects are `ImmutableDenseNDimArray`s; nested rendering, every scalar entry selectable.

```python
edit(Array([[x, y], [z, t]]))
```

```python
a3 = edit(Array([[[x, 1], [y, 2]], [[z, 3], [t, 4]]]))
a3
```

```python
a3.expr.shape, a3.expr[1, 0, 1]
```

```python
edit(tensorproduct(Array([x, y]), Array([1, z])))
```

```python
edit(derive_by_array(x**2 * y + sin(z), [x, y, z]))
```

## Sparse containers

```python
edit(SparseMatrix([[x, 0], [0, y]]))
```

## Notes

* Replacing an entry rebuilds the container with `func(*args)` (block matrices are rebuilt from their rows).  The shape is fixed by the container: to change the shape, edit the whole expression (select the root, Enter) and type e.g. `Matrix([[1, 2, 3]])`.
* Typed input is parsed with the expression's symbols in scope; `A`/`B` above resolve to the `MatrixSymbol`s, so `A*B + B` is valid input inside a matrix expression.
